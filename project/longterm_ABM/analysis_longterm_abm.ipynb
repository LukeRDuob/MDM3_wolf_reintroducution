```python
import pandas as pd
```

```python
df = pd.read_csv("weekly_model_multiple_runs.csv")
```

```python
df["Deer Killed per Pack per Week"] = (
    df["Weekly Deer Killed"] / df["Number of Packs"].clip(lower=1)
)

df["Deer Killed per Wolf per Week"] = (
    df["Weekly Deer Killed"] / df["Total Wolves"].clip(lower=1)
)
df["Deer Killed per Wolf per Year"] = df["Deer Killed per Wolf per Week"] * 52

df["Deer Births % pop per Week"] = (
    df["Weekly Deer Births"] / df["Total Deer"].clip(lower=1)
)
df["Deer Births % pop per Year"] = df["Deer Births % pop per Week"] * 52

df["Wolf Births per Pack per Week"] = (
    df["Weekly Wolf Births"] / df["Number of Packs"].clip(lower=1)
)
df["Wolf Births per Pack per Year"] = df["Wolf Births per Pack per Week"] * 52

df["Deer Weekly Death Rate Realised"] = (
    df["Weekly Deer Natural Deaths"] / df["Total Deer"].clip(lower=1)
)
df["Deer Annual Death Rate Realised"] = df["Deer Weekly Death Rate Realised"] * 52

df["Wolf Weekly Death Rate Realised"] = (
    df["Weekly Wolf Natural Deaths"] / df["Total Wolves"].clip(lower=1)
)
df["Wolf Annual Death Rate Realised"] = df["Wolf Weekly Death Rate Realised"] * 52

run_summary = df.groupby("Run").agg({
    "Deer Killed per Pack per Week": "mean",
    "Deer Killed per Wolf per Year": "mean",
    "Deer Births % pop per Year": "mean",
    "Wolf Births per Pack per Year": "mean",
    "Deer Annual Death Rate Realised": "mean",
    "Wolf Annual Death Rate Realised": "mean",
    "Total Deer": "mean",
    "Total Wolves": "mean",
    "Number of Herds": "mean",
    "Number of Packs": "mean",
})

print(run_summary)
print("\nOverall means across runs:")
print(run_summary.mean())
```
